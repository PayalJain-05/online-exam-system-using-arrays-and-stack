# **Project Outline: Online Examination System**

This notebook is structured to implement an Online Examination System, divided into four main sections or 'Persons', each handling distinct functionalities:

---

## **Person 1 - Question & Data Foundation**
*   **Objective:** Define the core data for the examination.
*   **Key Elements:**
    *   Exam Name
    *   Number of Questions
    *   Question IDs
    *   Question Texts
    *   Answer Options
    *   Correct Answers

---

## **Person 2 - Exam Attempt & Stack-Based Navigation**
*   **Objective:** Manage the student's exam attempt, including navigation and answer recording.
*   **Key Elements:**
    *   `Stack` Data Structure implementation (LIFO)
    *   Student Login (Name, ID)
    *   Student Answer Storage
    *   Question Navigation (Previous, Next)
    *   Mark for Review functionality (using a stack)
    *   Review Marked Questions functionality (using a stack)
    *   Exam Submission logic

---

## **Person 3 - Evaluation + Result**
*   **Objective:** Evaluate the student's performance and display their results.
*   **Key Elements:**
    *   Calculate Correct, Wrong, and Unattempted Answers
    *   Calculate Score and Percentage
    *   Display detailed results (Student Info, Score Breakdown)
    *   Categorize questions (Correct, Wrong, Unattempted)
    *   Provide Performance Feedback (e.g., "Excellent", "Needs Improvement")

---

## **Person 4 - Data Structures and Algorithms Module**
*   **Objective:** Implement and demonstrate core Data Structures and Algorithms related to the exam data, presented interactively.
*   **Key Elements:**
    *   **Question Search:** Implement a Linear Search algorithm to find specific questions.
    *   **Student Search:** Implement a Linear Search algorithm to find specific student results.
    *   **Exam Leaderboard:** Implement a Bubble Sort algorithm to rank students by score and display the leaderboard.
    *   **Algorithm Complexity Analysis:** Provide theoretical time and space complexity for implemented algorithms.
    *   **Interactive UI (ipywidgets):** Consolidate all Person 4 functionalities into a single, user-friendly interactive interface for easy access and demonstration.

```python
# ============================================================
# PERSON 1 - QUESTION & DATA FOUNDATION
# + EXAM INFORMATION UI
# ============================================================

import ipywidgets as widgets
from IPython.display import display, HTML


# ============================================================
# QUESTION & EXAM DATA
# ============================================================

exam_name = "Data Structures Online Examination"

number_of_questions = 5

question_ids = ["Q1", "Q2", "Q3", "Q4", "Q5"]

questions = [
    "Which data structure follows LIFO?",
    "Which data structure follows FIFO?",
    "What is an array?",
    "Which operation adds an element to a stack?",
    "Which operation removes an element from a stack?"
]

options = [
    ["A. Queue", "B. Stack", "C. Tree", "D. Graph"],

    ["A. Stack", "B. Queue", "C. Tree", "D. Graph"],

    [
        "A. A collection of elements stored in an ordered manner",
        "B. A type of loop",
        "C. A programming language",
        "D. An operating system"
    ],

    ["A. Pop", "B. Push", "C. Peek", "D. Delete"],

    ["A. Push", "B. Insert", "C. Pop", "D. Add"]
]

correct_answers = ["B", "B", "A", "B", "C"]


# ============================================================
# PERSON 1 UI
# ============================================================

person1_ui = widgets.VBox([

    widgets.HTML(
        f"""
        <div style="
            padding:25px;
            border-radius:16px;
            background:linear-gradient(135deg,#0d6efd,#6610f2);
            color:white;
            text-align:center;
            margin-bottom:15px;
        ">

            <h1 style="margin:0;">
                🎓 {exam_name}
            </h1>

            <p style="font-size:16px;">
                Online Student Examination Portal
            </p>

        </div>
        """
    ),

    widgets.HTML(
        f"""
        <div style="
            padding:20px;
            border:2px solid #0d6efd;
            border-radius:14px;
            background:#f8fbff;
        ">

            <h2 style="color:#0d6efd;">
                📋 Examination Details
            </h2>

            <p>
                <b>Subject:</b> Data Structures
            </p>

            <p>
                <b>Total Questions:</b> {number_of_questions}
            </p>

            <p>
                <b>Question IDs:</b>
                {", ".join(question_ids)}
            </p>

            <p>
                <b>Question Bank Status:</b>
                <span style="color:green;">
                    ✓ Loaded Successfully
                </span>
            </p>

        </div>
        """
    )

])

display(person1_ui)

print("✓ Person 1: Question and data foundation loaded.")
```

```python
# ============================================================
# PERSON 2 - EXAM ATTEMPT + STACK
# + STUDENT INTERACTIVE EXAM UI
# ============================================================

import ipywidgets as widgets
from IPython.display import display, clear_output, HTML


# ============================================================
# STACK CLASS
# ============================================================

class Stack:

    def __init__(self):
        self.stack = []

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        return None

    def peek(self):
        if not self.is_empty():
            return self.stack[-1]
        return None

    def is_empty(self):
        return len(self.stack) == 0

    def contains(self, item):
        return item in self.stack


# ============================================================
# EXAM STATE
# ============================================================

student_name = ""
student_id = ""

student_answers = [""] * number_of_questions

previous_stack = Stack()
review_stack = Stack()

current_question = 0

exam_started = False
exam_submitted = False


# ============================================================
# LOGIN WIDGETS
# ============================================================

student_name_input = widgets.Text(
    description="Name:",
    placeholder="Enter your name",
    layout=widgets.Layout(width="320px")
)

student_id_input = widgets.Text(
    description="ID:",
    placeholder="Enter Student ID",
    layout=widgets.Layout(width="320px")
)

start_exam_button = widgets.Button(
    description="Start Exam",
    button_style="success",
    icon="play",
    layout=widgets.Layout(width="160px")
)

login_message = widgets.HTML()


# ============================================================
# EXAM WIDGETS
# ============================================================

question_display = widgets.HTML()

options_radio = widgets.RadioButtons(
    options=[],
    description="",
    layout=widgets.Layout(width="100%")
)

progress_bar = widgets.IntProgress(
    value=0,
    min=0,
    max=number_of_questions,
    description="Progress:",
    layout=widgets.Layout(width="70%")
)

progress_label = widgets.Label()

status_message = widgets.HTML()


previous_button = widgets.Button(
    description="Previous",
    icon="arrow-left"
)

next_button = widgets.Button(
    description="Next",
    button_style="info",
    icon="arrow-right"
)

mark_button = widgets.Button(
    description="Mark for Review",
    button_style="warning",
    icon="bookmark"
)

review_button = widgets.Button(
    description="Review Marked",
    button_style="primary",
    icon="search"
)

submit_button = widgets.Button(
    description="Submit Exam",
    button_style="danger",
    icon="check"
)


# ============================================================
# EXAM AREA
# ============================================================

exam_area = widgets.VBox([
    widgets.HTML(
        """
        <h2 style="color:#198754;">
            📝 Attempt Your Examination
        </h2>
        """
    ),

    widgets.HTML(
        """
        <p>
            Select one option for each question.
            You can move backward, forward or mark questions
            for review.
        </p>
        """
    ),

    question_display,

    options_radio,

    widgets.HBox([
        progress_bar,
        progress_label
    ]),

    status_message,

    widgets.HBox(
        [
            previous_button,
            next_button,
            mark_button,
            review_button,
            submit_button
        ],
        layout=widgets.Layout(
            flex_flow="row wrap",
            gap="8px"
        )
    )
])

exam_area.layout.display = "none"


# ============================================================
# DISPLAY QUESTION
# ============================================================

def display_question():

    global current_question

    q = current_question

    question_display.value = f"""
    <div style="
        padding:22px;
        border:2px solid #0d6efd;
        border-radius:15px;
        background:#f8fbff;
        margin:15px 0;
    ">

        <h3 style="color:#0d6efd;">
            Question {q + 1} of {number_of_questions}
        </h3>

        <p>
            <b>Question ID:</b> {question_ids[q]}
        </p>

        <p style="
            font-size:19px;
            line-height:1.6;
        ">
            {questions[q]}
        </p>

    </div>
    """

    options_radio.options = options[q]

    options_radio.value = None

    # Restore saved answer
    if student_answers[q] != "":

        saved = student_answers[q]

        for option in options[q]:

            if option.startswith(saved + "."):

                options_radio.value = option
                break

    progress_bar.value = q + 1

    progress_label.value = (
        f"{q + 1}/{number_of_questions}"
    )


# ============================================================
# SAVE ANSWER
# ============================================================

def save_answer(change):

    if not exam_started:
        return

    if change["new"] is None:
        return

    selected = change["new"]

    student_answers[current_question] = selected[0]

    status_message.value = """
    <span style="color:#198754;font-weight:bold;">
        ✓ Answer saved
    </span>
    """


options_radio.observe(
    save_answer,
    names="value"
)


# ============================================================
# START EXAM
# ============================================================

def start_exam(button):

    global student_name
    global student_id
    global student_answers
    global previous_stack
    global review_stack
    global current_question
    global exam_started
    global exam_submitted

    student_name = student_name_input.value.strip()
    student_id = student_id_input.value.strip()

    if not student_name or not student_id:

        login_message.value = """
        <p style="color:#dc3545;">
            ⚠ Please enter your Name and Student ID.
        </p>
        """

        return

    student_answers = [""] * number_of_questions

    previous_stack = Stack()
    review_stack = Stack()

    current_question = 0

    exam_started = True
    exam_submitted = False

    student_name_input.disabled = True
    student_id_input.disabled = True
    start_exam_button.disabled = True

    login_message.value = f"""
    <div style="
        padding:12px;
        background:#e8f5e9;
        border-radius:10px;
        color:#198754;
    ">
        ✓ Welcome <b>{student_name}</b>.
        Your examination has started.
    </div>
    """

    exam_area.layout.display = "flex"

    display_question()


start_exam_button.on_click(start_exam)


# ============================================================
# NEXT QUESTION
# ============================================================

def next_question(button):

    global current_question

    if not exam_started:
        return

    if current_question < number_of_questions - 1:

        previous_stack.push(current_question)

        current_question += 1

        display_question()

    else:

        status_message.value = """
        <span style="color:#fd7e14;">
            ⚠ This is the last question.
            Please review your answers and submit.
        </span>
        """


next_button.on_click(next_question)


# ============================================================
# PREVIOUS QUESTION
# ============================================================

def previous_question(button):

    global current_question

    if not exam_started:
        return

    if not previous_stack.is_empty():

        current_question = previous_stack.pop()

        display_question()

    else:

        status_message.value = """
        <span style="color:#fd7e14;">
            ⚠ No previous question available.
        </span>
        """


previous_button.on_click(previous_question)


# ============================================================
# MARK FOR REVIEW
# ============================================================

def mark_for_review(button):

    if not exam_started:
        return

    if not review_stack.contains(current_question):

        review_stack.push(current_question)

        status_message.value = f"""
        <span style="color:#6f42c1;">
            🔖 Question {question_ids[current_question]}
            marked for review.
        </span>
        """

    else:

        status_message.value = """
        <span style="color:#fd7e14;">
            ⚠ Question already marked for review.
        </span>
        """


mark_button.on_click(mark_for_review)


# ============================================================
# REVIEW MARKED QUESTION
# ============================================================

def review_marked(button):

    global current_question

    if not exam_started:
        return

    if not review_stack.is_empty():

        previous_stack.push(current_question)

        current_question = review_stack.pop()

        display_question()

        status_message.value = """
        <span style="color:#6f42c1;">
            🔖 Opening marked question.
        </span>
        """

    else:

        status_message.value = """
        <span style="color:#fd7e14;">
            ⚠ No questions are marked for review.
        </span>
        """


review_button.on_click(review_marked)


# ============================================================
# SUBMIT
# ============================================================

def submit_exam(button):

    global exam_started
    global exam_submitted

    if not exam_started:
        return

    exam_started = False
    exam_submitted = True

    options_radio.disabled = True

    previous_button.disabled = True
    next_button.disabled = True
    mark_button.disabled = True
    review_button.disabled = True
    submit_button.disabled = True

    status_message.value = """
    <div style="
        padding:15px;
        background:#e8f5e9;
        border-radius:10px;
        color:#198754;
        font-weight:bold;
    ">
        ✓ Examination submitted successfully.
        <br>
        Proceed to the Person 3 result section.
    </div>
    """

    # Automatically evaluate if Person 3 has already been run
    if "evaluate_exam" in globals():

        evaluate_exam()


submit_button.on_click(submit_exam)


# ============================================================
# FINAL PERSON 2 UI
# ============================================================

person2_ui = widgets.VBox([

    widgets.HTML(
        f"""
        <div style="
            padding:20px;
            background:#f8fbff;
            border-radius:14px;
            border:2px solid #0d6efd;
        ">

            <h2 style="color:#0d6efd;">
                👤 Student Login
            </h2>

            <p>
                <b>Exam:</b> {exam_name}
            </p>

        </div>
        """
    ),

    widgets.HBox([
        student_name_input,
        student_id_input
    ]),

    start_exam_button,

    login_message,

    exam_area
])


display(person2_ui)

print("✓ Person 2: Student examination UI loaded.")
```

```python
# ============================================================
# PERSON 3 - EVALUATION + RESULT
# ============================================================

import ipywidgets as widgets
from IPython.display import display, HTML


# ============================================================
# RESULT VARIABLES
# ============================================================

correct_count = 0
wrong_count = 0
unattempted_count = 0

correct_questions = []
wrong_questions = []
unattempted_questions = []

score = 0
total_marks = number_of_questions
percentage = 0


# ============================================================
# RESULT UI
# ============================================================

result_output = widgets.HTML()


# ============================================================
# EVALUATE EXAM
# ============================================================

def evaluate_exam():

    global correct_count
    global wrong_count
    global unattempted_count

    global correct_questions
    global wrong_questions
    global unattempted_questions

    global score
    global percentage

    # Reset
    correct_count = 0
    wrong_count = 0
    unattempted_count = 0

    correct_questions = []
    wrong_questions = []
    unattempted_questions = []


    # ========================================================
    # CHECK ANSWERS
    # ========================================================

    for i in range(number_of_questions):

        student_answer = student_answers[i]

        correct_answer = correct_answers[i]

        if student_answer == correct_answer:

            correct_count += 1

            correct_questions.append(
                question_ids[i]
            )

        elif student_answer == "":

            unattempted_count += 1

            unattempted_questions.append(
                question_ids[i]
            )

        else:

            wrong_count += 1

            wrong_questions.append(
                question_ids[i]
            )


    # ========================================================
    # SCORE
    # ========================================================

    score = correct_count

    percentage = (
        score / total_marks
    ) * 100


    # ========================================================
    # PERFORMANCE
    # ========================================================

    if percentage >= 80:

        performance = "Excellent Performance! 🌟"
        performance_color = "#198754"

    elif percentage >= 60:

        performance = "Good Performance! 👍"
        performance_color = "#0d6efd"

    elif percentage >= 40:

        performance = "Average Performance."
        performance_color = "#fd7e14"

    else:

        performance = "Needs Improvement."
        performance_color = "#dc3545"


    # ========================================================
    # RESULT UI
    # ========================================================

    result_output.value = f"""

    <div style="
        margin-top:25px;
        padding:25px;
        border:2px solid #0d6efd;
        border-radius:18px;
        background:#f8fbff;
    ">

        <h1 style="
            text-align:center;
            color:#0d6efd;
        ">
            🎓 EXAM RESULT
        </h1>


        <div style="
            text-align:center;
            margin-bottom:20px;
        ">

            <h3>
                {student_name}
            </h3>

            <p>
                Student ID: <b>{student_id}</b>
            </p>

        </div>


        <!-- RESULT CARDS -->

        <div style="
            display:flex;
            gap:15px;
            flex-wrap:wrap;
        ">


            <div style="
                flex:1;
                min-width:140px;
                padding:18px;
                background:#e8f5e9;
                border-radius:12px;
                text-align:center;
            ">

                <b>✓ CORRECT</b>

                <h2>
                    {correct_count}
                </h2>

            </div>


            <div style="
                flex:1;
                min-width:140px;
                padding:18px;
                background:#ffebee;
                border-radius:12px;
                text-align:center;
            ">

                <b>✗ WRONG</b>

                <h2>
                    {wrong_count}
                </h2>

            </div>


            <div style="
                flex:1;
                min-width:140px;
                padding:18px;
                background:#fff3e0;
                border-radius:12px;
                text-align:center;
            ">

                <b>— UNATTEMPTED</b>

                <h2>
                    {unattempted_count}
                </h2>

            </div>

        </div>


        <!-- SCORE -->

        <div style="
            margin-top:20px;
            padding:20px;
            background:white;
            border-radius:14px;
            text-align:center;
            border:1px solid #ddd;
        ">

            <p>
                <b>YOUR SCORE</b>
            </p>

            <h1 style="
                font-size:40px;
                color:#0d6efd;
                margin:5px;
            ">
                {score} / {total_marks}
            </h1>

            <h2>
                {percentage:.2f}%
            </h2>

            <h3 style="
                color:{performance_color};
            ">
                {performance}
            </h3>

        </div>


        <!-- QUESTION BREAKDOWN -->

        <div style="
            margin-top:20px;
        ">

            <h3 style="color:#198754;">
                ✓ Correct Questions
            </h3>

            <p>
                {
                    ", ".join(correct_questions)
                    if correct_questions
                    else "None"
                }
            </p>


            <h3 style="color:#dc3545;">
                ✗ Wrong Questions
            </h3>

            <p>
                {
                    ", ".join(wrong_questions)
                    if wrong_questions
                    else "None"
                }
            </p>


            <h3 style="color:#fd7e14;">
                — Unattempted Questions
            </h3>

            <p>
                {
                    ", ".join(unattempted_questions)
                    if unattempted_questions
                    else "None"
                }
            </p>

        </div>

    </div>
    """


# ============================================================
# RESULT SECTION
# ============================================================

person3_ui = widgets.VBox([

    widgets.HTML(
        """
        <div style="
            margin-top:30px;
            padding:18px;
            border-left:6px solid #198754;
            background:#f8fff9;
            border-radius:10px;
        ">

            <h2 style="
                color:#198754;
                margin:0;
            ">
                📊 Evaluation & Result
            </h2>

            <p>
                Your examination answers are evaluated here.
            </p>

        </div>
        """
    ),

    result_output

])


display(person3_ui)

print("✓ Person 3: Evaluation system loaded.")
```

```python
# ============================================================
# PERSON 4 - DATA STRUCTURES & ALGORITHMS
# + SORTING / RANKING / LEADERBOARD UI
# ============================================================

import ipywidgets as widgets
from IPython.display import display, HTML


# ============================================================
# EXISTING STUDENT RESULT DATA
# ============================================================

students = [
    ["S101", "Payal", 78],
    ["S102", "Rahul", 92],
    ["S103", "Aman", 65],
    ["S104", "Riya", 88],
    ["S105", "Neha", 95]
]


# ============================================================
# LINEAR SEARCH - STUDENT
# ============================================================

def linear_search_student(student_id):

    for i in range(len(students)):

        if students[i][0].upper() == student_id.upper():

            return i

    return -1


# ============================================================
# LINEAR SEARCH - QUESTION
# Kept as Person 4's DAA implementation.
# NOT exposed as a student search screen.
# ============================================================

def linear_search_question(search_id):

    for i in range(number_of_questions):

        if question_ids[i].upper() == search_id.upper():

            return i

    return -1


# ============================================================
# BUBBLE SORT
# ============================================================

def bubble_sort_students(student_list):

    n = len(student_list)

    for i in range(n - 1):

        for j in range(n - i - 1):

            if student_list[j][2] < student_list[j + 1][2]:

                student_list[j], student_list[j + 1] = (
                    student_list[j + 1],
                    student_list[j]
                )

    return student_list


# ============================================================
# ADD CURRENT STUDENT AUTOMATICALLY
# ============================================================

def add_current_student_to_results():

    global students

    # Prevent duplicate addition
    existing = linear_search_student(student_id)

    if existing != -1:

        students[existing][2] = score

    else:

        students.append([
            student_id,
            student_name,
            score
        ])


# ============================================================
# PERSON 4 OUTPUT
# ============================================================

person4_output = widgets.HTML()


# ============================================================
# DISPLAY PERSON 4 RESULTS
# ============================================================

def display_person4_results():

    # Add current student's result
    add_current_student_to_results()


    # Sort using Bubble Sort
    sorted_students = bubble_sort_students(
        students.copy()
    )


    # Find current student's rank using Linear Search
    current_index = linear_search_student(student_id)

    current_rank = "N/A"

    if current_index != -1:

        current_score = students[current_index][2]

        for i, student in enumerate(sorted_students):

            if (
                student[0] == student_id
                and
                student[2] == current_score
            ):

                current_rank = i + 1

                break


    # ========================================================
    # LEADERBOARD HTML
    # ========================================================

    leaderboard_rows = ""

    for i, student in enumerate(sorted_students):

        rank = i + 1

        sid = student[0]
        name = student[1]
        marks = student[2]

        if sid == student_id:

            leaderboard_rows += f"""
            <tr style="
                background:#e7f1ff;
                font-weight:bold;
            ">
                <td>{rank}</td>
                <td>{sid}</td>
                <td>{name}</td>
                <td>{marks}</td>
            </tr>
            """

        else:

            leaderboard_rows += f"""
            <tr>
                <td>{rank}</td>
                <td>{sid}</td>
                <td>{name}</td>
                <td>{marks}</td>
            </tr>
            """


    # ========================================================
    # ALGORITHM ANALYSIS
    # ========================================================

    person4_output.value = f"""

    <div style="
        margin-top:30px;
        padding:25px;
        border:2px solid #6610f2;
        border-radius:18px;
        background:#faf8ff;
    ">

        <h1 style="
            text-align:center;
            color:#6610f2;
        ">
            🏆 PERFORMANCE & RANKING
        </h1>


        <!-- CURRENT STUDENT -->

        <div style="
            padding:20px;
            background:white;
            border-radius:14px;
            border:1px solid #ddd;
        ">

            <h2 style="color:#6610f2;">
                Your Performance
            </h2>

            <p>
                <b>Student:</b>
                {student_name}
            </p>

            <p>
                <b>Student ID:</b>
                {student_id}
            </p>

            <p>
                <b>Score:</b>
                {score}/{total_marks}
            </p>

            <p>
                <b>Percentage:</b>
                {percentage:.2f}%
            </p>

            <p style="
                font-size:20px;
                color:#0d6efd;
                font-weight:bold;
            ">
                🏅 Your Rank: {current_rank}
            </p>

        </div>


        <!-- LEADERBOARD -->

        <div style="
            margin-top:25px;
        ">

            <h2 style="color:#6610f2;">
                🏆 Exam Leaderboard
            </h2>

            <table style="
                width:100%;
                border-collapse:collapse;
                background:white;
            ">

                <tr style="
                    background:#6610f2;
                    color:white;
                ">

                    <th style="padding:10px;">
                        Rank
                    </th>

                    <th style="padding:10px;">
                        Student ID
                    </th>

                    <th style="padding:10px;">
                        Student
                    </th>

                    <th style="padding:10px;">
                        Score
                    </th>

                </tr>

                {leaderboard_rows}

            </table>

        </div>


        <!-- ALGORITHM ANALYSIS -->

        <div style="
            margin-top:25px;
            padding:20px;
            background:#f8f9fa;
            border-radius:14px;
        ">

            <h2 style="color:#6610f2;">
                ⚙️ Algorithm Analysis
            </h2>

            <p>
                <b>Question Linear Search</b>
            </p>

            <p>
                Best Case: O(1)
                <br>
                Worst Case: O(n)
                <br>
                Space: O(1)
            </p>

            <hr>

            <p>
                <b>Student Linear Search</b>
            </p>

            <p>
                Best Case: O(1)
                <br>
                Worst Case: O(n)
                <br>
                Space: O(1)
            </p>

            <hr>

            <p>
                <b>Bubble Sort</b>
            </p>

            <p>
                Best Case: O(n²)
                <br>
                Worst Case: O(n²)
                <br>
                Space: O(1)
            </p>

        </div>

    </div>
    """


# ============================================================
# BUTTON
# ============================================================

show_results_button = widgets.Button(
    description="View Ranking & Performance",
    button_style="primary",
    icon="trophy"
)


def show_person4(button):

    # Make sure exam has actually been evaluated
    if "score" not in globals():

        person4_output.value = """
        <div style="
            padding:20px;
            background:#fff3f3;
            border:2px solid #dc3545;
            border-radius:12px;
        ">

            <h3 style="color:#dc3545;">
                ⚠ Please complete the examination first.
            </h3>

        </div>
        """

        return

    display_person4_results()


show_results_button.on_click(show_person4)


# ============================================================
# PERSON 4 UI
# ============================================================

person4_ui = widgets.VBox([

    widgets.HTML(
        """
        <div style="
            margin-top:30px;
            padding:18px;
            border-left:6px solid #6610f2;
            background:#faf8ff;
            border-radius:10px;
        ">

            <h2 style="
                margin:0;
                color:#6610f2;
            ">
                🏆 Ranking & DAA Analysis
            </h2>

            <p>
                Your result is now processed using
                searching and sorting algorithms.
            </p>

        </div>
        """
    ),

    show_results_button,

    person4_output

])


display(person4_ui)

print("✓ Person 4: DAA and ranking UI loaded.")
```
